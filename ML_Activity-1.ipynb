The missing-value analysis shows that most columns in the Titanic dataset have no missing values. The `Age` column contains 177 missing values, representing approximately 19.87% of the dataset. The `Cabin` column has 687 missing values, representing approximately 77.10%, which is a substantial amount of missing data. The `Embarked` column has only 2 missing values, representing approximately 0.22%.

Therefore, `Age`, `Cabin`, and `Embarked` require attention during the data-cleaning process. The remaining columns do not contain missing values.

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Clean dataset shape:", df_clean.shape)
```

```python
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())
df_clean['Embarked'] = df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0])
df_clean = df_clean.drop(columns=['Cabin'])
```

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

The original dataset was copied before performing any cleaning operations. Missing values in `Age` were replaced using the median because it is a numerical variable and the median is less sensitive to extreme values. Missing values in `Embarked` were replaced using the mode because it is a categorical variable and only two values were missing.
The `Cabin` column was removed because approximately 77% of its values were missing, making reliable imputation difficult. After handling the missing values, the dataset was checked again and no missing values remained.

```python
categorical_columns = df_clean.select_dtypes(include=['object']).columns.tolist()
print("Categorical columns:")
print(categorical_columns)
print("\nColumns selected for encoding:")
print(['Sex', 'Embarked'])
```

```python
df_encoded = pd.get_dummies(
    df_clean,
    columns=['Sex', 'Embarked'],
    drop_first=True,
    dtype=int
)

print("Categorical encoding completed.")
```

```python
print("Columns before encoding:")
print(df_clean.columns.tolist())
print("\nColumns after encoding:")
print(df_encoded.columns.tolist())
```

The categorical columns `Sex`, `Embarked`, `Name`, and `Ticket` were identified. For machine learning representation, `Sex` and `Embarked` were selected for encoding because they contain meaningful categorical information with no natural numerical order.
One-hot encoding was applied to `Sex` and `Embarked`. The `Sex` column was converted into `Sex_male`, where 0 represents female and 1 represents male. The `Embarked` column was converted into dummy variables representing the different embarkation ports.
`drop_first=True` was used to remove one category from each feature and avoid redundant information.
`Name` and `Ticket` were not encoded because they contain a large number of unique values and direct encoding would create many unnecessary features. `PassengerId` is an identifier and does not require encoding.
No label encoding was used because the selected categorical variables (`Sex` and `Embarked`) are nominal categories without a natural order.

```python
numerical_columns = df_encoded.select_dtypes(include=['int64', 'float64']).columns
print("Numerical columns:")
print(numerical_columns.tolist())
print("\nMinimum and Maximum values:")
print(df_encoded[numerical_columns].agg(['min', 'max']).T)
```

```python
from sklearn.preprocessing import StandardScaler
scaling_columns = ['Age', 'SibSp', 'Parch', 'Fare']
scaler = StandardScaler()
df_encoded[scaling_columns] = scaler.fit_transform(
    df_encoded[scaling_columns]
)
print("Standardisation completed successfully.")
```

```python

print("Scaled features:")
print(df_encoded[scaling_columns].head())
print("\nMean after scaling:")
print(df_encoded[scaling_columns].mean().round(2))
print("\nStandard deviation after scaling:")
print(df_encoded[scaling_columns].std().round(2))
```

*The numerical features `Age`, `SibSp`, `Parch`, and `Fare` have different measurement ranges. Scaling can be important for machine learning algorithms that are sensitive to feature magnitude, such as K-Nearest Neighbors, Support Vector Machines, and algorithms that use gradient-based optimisation.
*Standardisation using `StandardScaler` was selected. It transforms the selected numerical features so that they have approximately a mean of 0 and a standard deviation of 1.
*`PassengerId` was not scaled because it is an identifier. `Survived` was not scaled because it is the target variable for a potential future prediction task. `Pclass` was also kept unchanged because it represents an ordinal passenger class.
*In a predictive machine learning workflow, the scaler should be fitted only on the training data and then used to transform the validation or test data. In this assignment, scaling is demonstrated on the prepared dataset because no predictive model is being trained.

# Conclusion

The Titanic dataset was selected as an open-source transport dataset for this data preparation task. The dataset originally contained 891 rows and 12 columns.

During data quality analysis, missing values were found in the `Age`, `Cabin`, and `Embarked` columns. Missing `Age` values were replaced using the median, while missing `Embarked` values were replaced using the mode. The `Cabin` column was removed because approximately 77% of its values were missing.

Duplicate records were checked and no unwanted duplicate rows were found. Categorical values were also inspected for inconsistent labels or extra spaces. The `Sex` and `Embarked` features were converted using one-hot encoding because they are nominal categorical variables.

For numerical features, `Age`, `SibSp`, `Parch`, and `Fare` were standardised using `StandardScaler`. The `PassengerId` identifier and `Survived` target variable were not scaled.

Finally, the prepared dataset was checked for remaining missing values and duplicate records and was saved separately as `titanic_prepared.csv`. The original dataset remained unchanged.
